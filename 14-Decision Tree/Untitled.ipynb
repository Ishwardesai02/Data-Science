```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
data=pd.read_csv("credit.csv")
data.head()
```

```python
##data preparation
#check for null values
data.isnull().sum()
```

```python
data.dropna()
```

```python
data.columns
```

```python
#there are 9 columns having nan nyumeric values , let us 
#there is one column called phone which is not useful
data=data.drop(["phone"],axis=1)
```

```python
#NOw there are 16 coulumns 
lb=LabelEncoder()
data["Checking_balance"]=lb.fit_transform(data["checking_balance"])
data["credit_history"]=lb.fit_transform(data["credit_history"])
data["purpose"]=lb.fit_transform(data["purpose"])
data["savings_balance"]=lb.fit_transform(data["savings_balance"])
data["employment_duration"]=lb.fit_transform(data["employment_duration"])
data["other_credit"]=lb.fit_transform(data["other_credit"])
data["housing"]=lb.fit_transform(data["housing"])
data["job"]=lb.fit_transform(data["job"])
```

```python
##check the non numeric columns
non_numeric_cols=data.select_dtypes(include=["object"]).columns
print(non_numeric_cols)
```

```python
data["Checking_balance"]=lb.fit_transform(data["checking_balance"])
data["default"]=lb.fit_transform(data["default"])
```

```python
##now let us check how many unique values are there is target column
data["default"].unique()
```

```python
data["default"].value_counts()
```

```python
#now we want to split tree , we need all fetaures coilumns
colnames=list(data.columns)
```

```python
#Now lwt us assign these columns to variable predictor 
predictors=colnames[:15]
target=colnames[15]
```

```python
#splitting data into training and testing data set
from sklearn.model_selection import train_test_split
train,test=train_test_split(data,test_size=0.3)
```

```python
from sklearn.tree import DecisionTreeClassifier as DT
help(DT)

```

```python
model=DT(criterion='entropy')
model.fit(train[predictors],train[target])
```

```python
#Prediction on test data
preds=model.predict(test[predictors])
pd.crosstab(test[target],preds,rownames=['Actual'],colnames=['Predictions'])
```

```python
np.mean(preds==test[target]) #test data sacuuracy
```
