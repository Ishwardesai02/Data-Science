```python
import pandas as pd
from sklearn.ensemble import  AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn import metrics
import warnings
warnings.filterwarnings('ignore')
```

```python
#read csv files 
loan_data=pd.read_csv("income.csv")
loan_data.columns
loan_data.head()
```

```python
#Let us split the data in input and output
X=loan_data.iloc[:,0:6]
y=loan_data.iloc[:,6]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
#Create adaboost classifier
ada_model=AdaBoostClassifier(n_estimators=100,learning_rate=1)
#n_estimators=number of week learnenrs
#learning rates it contibutes weights of the weak learners,bydefault
```

```python
#train the model
model=ada_model.fit(X_train,y_train)
#Predict the results
y_pred=model.predict(X_test)
print("accuracy",metrics.accuracy_score(y_test,y_pred))
```

```python
#Let us try for another base model
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
#here base model is changes
Ada_model=AdaBoostClassifier(n_estimators=50,estimator=lr,learning_rate=1)
model=Ada_model.fit(X_train,y_train)
y_pred=model.predict(X_test)
print("Accuracy",metrics.accuracy_score(y_test,y_pred))
```
