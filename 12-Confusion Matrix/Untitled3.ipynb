```python
import pandas as pd
import numpy as np
df=pd.read_csv("wbcd.csv")
df.head()
```

```python
df.describe()
```

```python
df=df.iloc[:,1:32]
```

```python
def norm_fun(i):
    x=(i-i.max())/(i.max()-i.min())
    return x

df_n=norm_fun(df.iloc[:,1:32])
```

```python
df_n.head()
```

```python
#Let us now apply x as input and y asoutput
X=np.array(df_n.iloc[:,:])
##Since is wbcd norm we are alredy excluding output column ,hence all rows and 
y=np.array(df['diagnosis'])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.neighbors import KNeighborsClassifier 
knn=KNeighborsClassifier(n_neighbors=9)
knn.fit(X_train,y_train)
pred=knn.predict(X_test)
pred
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(pred,y_test)) 
pd.crosstab(pred,y_test)
```
