```python
import pandas as pd
import numpy as np
letters=pd.read_csv("letterdata.csv")
letters.head()
```

```python
"""Data set typically contains handwritten chareacters
letter :represents the target class
Features: this are numeric attributes describing various geometric or statastical properties of the chgaracters
xbox and ybox: x and y bounding box dimensions
width and height : width and height of the character bounding box
onpix:number of on-pixels in characters image
xbar and ybar : Mean X and Y coordib=nate valuesof on-pixels
x2bar,y2bar,and xybar : variance and covarience of pixel intensities
x2ybar and xy2bar : additional"""
```

```python
#let us carry out EDA
letters.describe()
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
train,test=train_test_split(letters,test_size=0.2)
```

```python
#Let us split the data in terms of X and Y for both train and test data
train_X=train.iloc[:,1:]
train_y=train.iloc[:,0]
test_X=test.iloc[:,1:]
test_y=test.iloc[:,0]
```

```python
#Kernel linear 
model_linear=SVC(kernel="linear")
model_linear.fit(train_X,train_y)
pred_test_linear=model_linear.predict(test_X)
```

```python
#now let us check the acuracy
np.mean(pred_test_linear==test_y)
```

```python
#Kernel rbf
#Kernel linear 
model_rbf=SVC(kernel="rbf")
model_rbf.fit(train_X,train_y)
pred_test_rbf=model_rbf.predict(test_X)
```

```python
#now let us check the acuracy
np.mean(pred_test_rbf==test_y)
```
