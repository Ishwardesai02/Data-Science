```python
#Mann-Whitney Test
fuel=pd.read_csv("mann_whitney_additive.csv")
fuel.columns=["Without_addictive","With_addictive"]
```

```python
#Objective: To determine if there is a significant difference between the two promotional offers
offers=pd.read_excel('Promotion.xlsx')
offers.columns=['InterestRateWaiver','StandardPromotion']
offers.head()
```

```python
#Normality Tests
print("InterestRateWaiver normality: ", stats.shapiro(offers.InterestRateWaiver))
print("StandardPromotion normality: ", stats.shapiro(offers.StandardPromotion))
```

```python
#Variance test
levene_test=scipy.stats.levene(offers.InterestRateWaiver,offers.StandardPromotion)
print("Levene's test results: ",levene_test)
#p value=0.2875
#H0: Variance equal
#H1: Variance not equal
#p value>0.05 fail to reject null hypothesis
```

```python
#Two sample t test
ttest_result=stats.ttest_ind(offers.InterestRateWaiver,offers.StandardPromotion)
print("Two Sample T-test",ttest_result)
#Result : p-value=0.02422
#Interpretation
#Ho both offers have the same mean impoact
#H1 the mean impacts of both offers are diffrent
##Since the p value (0.02422584468584315) is less then 0.05 ,we reject null hypothesis
#Conclusiuon:There is a significance diffrence between two promotional offers
```

```python
#Moods Median Test
#Objective : is the medians of piglet,pooh, and tigger are statistacally equal
#It has equal medians or not
animals=pd.read_csv("animals.csv")
```

```python
#Normality Tests
print("Pooh Normality :",stats.shapiro(animals.Pooh))
#pvalue=0.0122
print("Piglet Noramality:",stats.shapiro(animals.Piglet))
#pvalue=0.044
print("Tigger Normality :",stats.shapiro(animals.Tigger))
#pvalue=0.0219
#H0 data is normal 
#H1 data is not normal
##Since all the pvalues arew less then 0.05 hence reject the null hypothesis
#Data is not normal hence Moods test
```

```python
#Median test
median_test_result=stats.median_test(animals.Pooh,animals.Piglet,animals.Tigger)
print("Mood's Median test",median_test_result)
#Result:pvalue=>0.186
#Interpretation
#H0:All groups have equal medians
#H1:All groups have significant diffrence between median
```

```python
#One way Anova test
#objective : is the transaction time for the three suppliers are not significantly diffrent 
contract=pd.read_excel("ContractRenewal_Data(unstacked).xlsx")
contract.columns=["Supp_A","Supp_B","Supp_C"]

```

```python
#Simlarly , if 720 chioldren were survyed and 152 of them 
#reported consuming soft drinks ,the second count is 152
#Thus count =[58,152]

#nobs = represent total number of individual suryved in each group

#The total number of adults survyed in 480
#the total number of children survyed in 740
#Hence nobs=[480,740]

#These values are often extracted from a dataset 
#If your data is in a file ("like "JohnyTalkers.xlsv"")
#You can calculate the values as follows

import pandas as pd
#load the dataset
filepath="JohnyTalkers.xlsx" #replace with your
soft_drink_data=pd.read_excel(filepath)

#Filter the data into sdults and chilren category 
adults=soft_drink_data[soft_drink_data["Person"]=="Adults"]
children=soft_drink_data[soft_drink_data["Person"]=="Children"]

#Count of success (soft drink consumers) for eachgroup
count_adults=adults[adults["Drinks"]=='Purchased'].shape[0]
count_children=children[children["Drinks"]=='Purchased'].shape[0]


#Total observations for each group
nobs_adults=adults.shape[0]
nobs_children=children.shape[0]

#Final arraya for Z-test
count=[count_adults,count_children]
nobs=[nobs_adults,nobs_children]

print("Counts (Soft Drink Consumers):",count)
print("Total observations:",nobs)
```

```python
#Two sided test
z_stat,z_value=proportions_ztest(count,nobs,alternative="two-sided")
print("Two-Sided Proportion Test:",z_stat,f"P-Value:{p_val:.7f}")
#Result:
#p value=0.000
#interpretation
#H0: Proportions of adults and children consuming the soft drinks are the same
#H1: Proportions are different
#p<0.05, reject H0
#Conclusion: proportions are different
```
